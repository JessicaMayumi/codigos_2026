## **Análise Exploratória de Dados**
- Data: 10/08/2026

```python
dados = {
    'Modelo': ["Mobi", "Kwid", "Golf", "Golf", "Fusca"],
    'ano': [2023, 2025, 2005, 2005, 1970],
    'PREÇO': [60000, 82000, 25000, 25000, 30000],
    'Cor': ["Preto", "Laranja", "PRETO", "Preto", "Verde"],
    'Km': [75000, 19000, 350000, 350000, 275000],
}
```

```python
import pandas as pd

df = pd.DataFrame(dados)
print(df)
```

```python
preco = df['PREÇO'] # Para quando eu quero capturar apenas uma coluna
print("Coluna PREÇO: ", preco)
print("Média da coluna: ", preco.mean())

print()
preco = df[df['PREÇO'] < 50000] # 2 dfs pq a saída tmb eh um df || < 50000 eh um filtro no df inicial
print(preco)
```

#### Carros pretos com km < 100k

```python
filtro1 = df[(df['Cor'].str.strip().str.upper() == 'PRETO') & (df['Km'] <100000)]
print(filtro1)
```

#### Nova coluna, quantos km o carro andou por ano

```python
import datetime as dt
anos = dt.datetime.now().year - df['ano']
df['KmAno'] = (df['Km'] / anos).round(2)
print(df)
```

#### Remover as duplicatas a partir de uma coluna (ex: cpf, id)

```python
df = df.drop_duplicates(subset=['Modelo']) # exclui sempre a última informação duplicada, e mantém a primeira
print(df)
```

#### Preencher valroes vazios

dropna apaga linhas q tem valores vazios

```python
import numpy as np
dados2 = {
    'Modelo': ["Mobi", "Kwid", "Golf", "Golf", "Fusca"],
    'ano': [2023, 2025, 2005, 2005, 1970],
    'PREÇO': [60000, 82000, 25000, 25000, 30000],
    'Cor': ["Preto", "Laranja", "PRETO", "Preto", "Verde"],
    'Km': [75000, np.nan, 350000, 350000, 275000],
}

df2 = pd.DataFrame(dados2)
print(df2)
```

```python
df2['Km'] = df2['Km'].fillna(df2['Km'].mean())
print(df2)
```

```python
df2['Km'] = df2['Km'].fillna(0)
print(df2)
```

#### Ordenação
- Ordenando do menor para o maior preço

```python
df_ordenado = df2.sort_values(by='PREÇO', ascending=True)
print(df_ordenado)
```

#### Renomear colunas

```python
df2 = df2.rename(columns={'Modelo':'MODELO', 'ano':'ANO', 'PREÇO': 'PRECO', 'Cor':'COR', 'Km':'KM'})
print(df2.sort_values(by='PRECO', ascending=True))
```

#### Exportar

```python
df2.to_csv('dados_tratados.csv', index=False)
```
